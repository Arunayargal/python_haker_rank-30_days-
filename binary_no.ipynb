```python
#!/bin/python3

import math
import os
import random
import re
import sys
if __name__ == '__main__':
    n = int(input().strip())
    binary = bin(n)[2:]      # Convert to binary and remove '0b'

    max_count = 0
    count = 0

    for digit in binary:
        if digit == '1':
            count += 1
            if count > max_count:
                max_count = count
        else:
            count = 0

    print(max_count)
```
